# B4: How much labelled data?

Training sizes 128 to 16,384 rows, nested stratified subsamples of each outer-training fold (at least one example per class; classes are never dropped), test rows unchanged; 5 draws up to 512 rows, 3 at 1,024–2,048, 1 above. Main curve: all 18 datasets at 128–2,048; extended curve: the seven datasets with at least 16,384 training rows at every size. Arms: baseline, each decision model's both view, frozen embeddings (per-column PCA32, three encoders) and the label-trained classifier on E5-small-v2 (per column); PCA and classifier fit only on the sampled rows. This deliberately subsamples training rows: an explicit, recorded, owner-approved design, not a silent cap. Common protocol: seven fixed predictors with library defaults and no tuning (linear/logistic, CatBoost, LightGBM, TabPFN-3.5, TabICL v2, LimiX-2, Kumo); the baseline uses the non-text columns only; 18 datasets with the same five outer folds. Fixed headline metrics: binary ROC-AUC, multiclass macro one-vs-rest ROC-AUC, regression R². Datasets are the unit: per-dataset paired gain, equal-dataset means with dataset-bootstrap intervals, win/tie/loss counts, raw gain and share of headroom. 

Tables come from `output_JEVPFN/exp11_data_efficiency/results/group_b/b4_data_efficiency.zip`.
