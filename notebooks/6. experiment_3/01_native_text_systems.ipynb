## 1. Scope and provenance

```python
import sys
from pathlib import Path
REPO_ROOT = next(p for p in (Path.cwd(), *Path.cwd().parents) if (p / "pyproject.toml").is_file() and (p / "src").is_dir())
if str(REPO_ROOT) not in sys.path:
    sys.path.insert(0, str(REPO_ROOT))
import matplotlib.pyplot as plt
from src.utils.notebook_display import display_local, display_json, display_table
from src.utils.notebook_report import begin_report, report_section, record, finish_report
from src.utils.provenance import provenance
from src.visualize import figures, style
style.apply()
NOTEBOOK = '6. experiment_3/01_native_text_systems'
begin_report()
save = figures.FigureSaver(NOTEBOOK)
report_section("1. Scope, provenance and saved evidence")
display_json("Notebook execution identity", provenance())
from src.evaluation.results import collect
from src.evaluation.notebook_study import planned_coverage, headline_summary, burden_table, contrast_summary
from src.visualize.study import coverage_figures, native_figures, burden_figures
from src.evaluation.endpoints import HEADLINES
status, folds, per_dataset, paired = collect()
summary = headline_summary(folds)
display_json("Saved result coverage", status)
display_json("Prespecified headline metrics", HEADLINES)
print("Only saved, validated folds are read. No fit, API call, download or hypothesis test runs here.")
```

## 2. Completion and model identity

One configuration per dataset and native system; the same five outer folds as Experiments 1 and 2.

```python
report_section('2. Completion and model identity')
coverage = planned_coverage(folds, "experiment_3")
display_table("Original-text configuration coverage", coverage, preview_rows=12)
for fig, name, caption in coverage_figures(coverage, "Experiment 3 configuration coverage"):
    save(fig, name, caption=caption)
    display_local(fig)
    plt.close(fig)
```

## 3. AUC and R-squared by dataset

Absolute task-specific metrics remain separate. Missing systems are not zero scores; negative R-squared values remain visible.

```python
report_section('3. AUC and R-squared by dataset')
native = summary[summary.phase.eq("experiment_3")]
display_table("Native system metrics, checkpoint identities and complete folds", native)
for fig, name, caption in native_figures(summary):
    save(fig, name, caption=caption)
    display_local(fig)
    plt.close(fig)
```

## 4. Training burden and text coverage

Native systems have different text windows and training procedures. Their differences cannot isolate a representation while holding the predictor fixed.

```python
report_section('4. Training burden and text coverage')
display_table("Native system fit/predict time and raw representation width", burden_table(folds, ["experiment_3"]))
record("TabSTAR v1.1 uses fixed downstream LoRA adaptation; ConTextTab uses semantic in-context learning; AG-MM uses fixed supervised text/tabular fusion. Publisher text limits and observed resources belong with the results. There is no HPO campaign.", "Reference-system interpretation")
```
