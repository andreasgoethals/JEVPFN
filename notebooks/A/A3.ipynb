# A3: Decision geometry versus frozen encoders (appendix)

**Question (A3).** Do texts that are close in frozen-encoder space also receive similar decision probabilities? Per dataset and text field, up to 500 distinct non-empty texts and all their pairs (per-column requests); raw encoder vectors before PCA (E5-small, E5-large, Qwen3-Embedding-0.6B) give cosine distances, decision probabilities give total variation distances. Spearman correlation of the two distance vectors and k-nearest-neighbour overlap, each with a permutation or chance reference. No probe or decoder is fitted; this analysis is descriptive only.

All tables come from **one compact export**, `output_JEVPFN/A/A3/results/geometry/<cohort-and-roster>/geometry.zip` (schema `jevpfn.rq1c.export.v2`), built by `scripts/slurm/A3_export.slurm`.

```python
import sys
from pathlib import Path

REPO_ROOT = next(p for p in (Path.cwd(), *Path.cwd().parents) if (p / "pyproject.toml").is_file() and (p / "src").is_dir())
if str(REPO_ROOT) not in sys.path:
    sys.path.insert(0, str(REPO_ROOT))
from src.evaluation import group_a, model_properties, study_export, study_reports
from src.utils.notebook_display import display_json, display_table
from src.utils.notebook_report import begin_report, finish_report, record, report_section
from src.utils.provenance import provenance
from src.visualize import figures, style
from src.visualize.geometry import geometry_figure
from src.visualize.group_b import render

style.apply()
NOTEBOOK = "A/A3"
begin_report()
from src.evaluation.benchmark_export import cohort_record
cohort = cohort_record()
ROSTER = cohort["decision_roster"]
record(cohort, "Decision-model cohort and members")
save = figures.FigureSaver(NOTEBOOK)
report_section("1. What is complete")
display_json("Notebook execution identity", provenance())
frames, receipt = study_reports.geometry_export_tables()
frames = group_a.cohort_frames(frames, ROSTER)
print(study_export.describe(receipt))
record({k: receipt.get(k) for k in ("status", "path", "created_utc", "origin", "sha256", "bytes", "schema", "receipt_sha256")}, "Export identity")
properties = model_properties.profile_table().loc[lambda t: t.provider.isin(ROSTER)]
order = group_a.model_order(properties.provider, properties)
if frames:
    display_table("Sampled fields and missing providers", frames["coverage"], preview_rows=8)
    record(frames["coverage"], "Geometry coverage")
```

## 2. Summary: distance correlation

Per decision model and encoder: Spearman correlation between encoder and decision distances over all text pairs, fields averaged within a dataset and datasets with equal weight. Intervals resample datasets (pairs are not independent). The shaded band is the permutation null (text identities permuted within a field); the lower panel gives the agreement between two encoders on the same samples as a reference scale.

```python
report_section("2. Summary: distance correlation")
if not frames:
    print("Waiting for export: no geometry tables are available. Each field contributes up to 500 distinct texts.")
else:
    render(save, geometry_figure(frames["comparisons"], frames["encoder_references"], metric="spearman", order=order))
    summary = group_a.geometry_summary(frames["comparisons"], "spearman")
    display_table("Distance correlation per model and encoder (equal dataset weight, dataset-bootstrap 95% CI)", summary.round(3), preview_rows=12)
    display_table("Encoder-encoder reference", group_a.encoder_reference_summary(frames["encoder_references"]).round(3))
    record(summary, "Geometry: distance correlation, equal dataset weight")
    record(group_a.geometry_by_dataset(frames["comparisons"]), "Geometry per dataset, encoder and model")
```

## 3. Neighbour overlap (supporting)

The same layout for the tie-aware overlap of each text's k nearest neighbours (k = min(10, (n − 1)/2)) in encoder and decision space, against the chance overlap k/(n − 1).

```python
report_section("3. Neighbour overlap")
if not frames:
    print("Waiting for export: no geometry tables are available.")
else:
    render(save, geometry_figure(frames["comparisons"], frames["encoder_references"], metric="neighbour_overlap", order=order))
    overlap = group_a.geometry_summary(frames["comparisons"], "neighbour_overlap")
    display_table("Neighbour overlap per model and encoder (equal dataset weight)", overlap.round(3), preview_rows=12)
    record(overlap, "Geometry: neighbour overlap, equal dataset weight")
```
