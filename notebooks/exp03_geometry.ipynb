# A3: Similarity structure versus frozen encoders

Do texts that are close in frozen-encoder space also receive similar decision probabilities? Per dataset and text field, up to 500 distinct non-empty texts and all their pairs; raw encoder vectors before PCA, cosine distance against the decision models' total variation, Spearman correlation and neighbourhood agreement with a permutation reference. No probe or decoder is fitted.

All tables come from **one compact export**, `output_JEVPFN/exp03_geometry/results/geometry/geometry.zip`, built on VSC by the Mindwell CPU job `scripts/slurm/rq1_geometry_export.slurm` (`python -m src.evaluation.geometry_export`) from the full geometry summaries on PROJECT.

```python
import sys
from pathlib import Path
REPO_ROOT = next(p for p in (Path.cwd(), *Path.cwd().parents) if (p / "pyproject.toml").is_file() and (p / "src").is_dir())
if str(REPO_ROOT) not in sys.path:
    sys.path.insert(0, str(REPO_ROOT))
from src.utils.notebook_display import display_json, display_table
from src.utils.notebook_report import begin_report, report_section, record, finish_report
from src.utils.provenance import provenance
from src.evaluation import study_export, study_reports
from src.visualize import figures, style
style.apply()
NOTEBOOK = "exp03_geometry"
begin_report()
save = figures.FigureSaver(NOTEBOOK)
report_section("1. Scope, provenance and the export read by this notebook")
display_json("Notebook execution identity", provenance())
frames, receipt = study_reports.geometry_export_tables()
print(study_export.describe(receipt))
record({k: receipt.get(k) for k in ("status", "path", "created_utc", "origin", "sha256", "bytes", "schema", "receipt_sha256")}, "Export identity")
```

```python
report_section("2. Do neighbours in embedding space receive similar decisions?")
if not frames:
    print("Waiting for export: no geometry tables are available. Each field contributes up to 500 distinct texts.")
else:
    relations, coverage = frames["comparisons"], frames["coverage"]
    display_table("Sampled fields and missing providers", coverage, preview_rows=8)
    display_table("Reference distances between frozen encoders on the same field samples", frames["encoder_references"], preview_rows=8)
    display_table("Observed similarity, permuted reference and neighbourhood overlap", relations, preview_rows=8)
    by_dataset = relations.groupby(["encoder", "provider", "dataset"])[["spearman", "null_mean", "neighbour_overlap"]].mean()
    summary = by_dataset.groupby(["encoder", "provider"]).mean()
    display_table("Equal-weight dataset means across observed fields", summary.round(3).reset_index(), preview_rows=12)
    display_table("Datasets with defined distance correlations", by_dataset.spearman.groupby(["encoder", "provider"]).count().to_frame("datasets").reset_index())
    record(summary.reset_index(), "Geometry: equal-weight dataset means")
```
