# B3: When and for whom do decision features help?

Four views, all descriptive and observational: **datasets** (label-free properties including decision redundancy — how well the non-text columns predict a model's joint decision output — with one Benjamini-Hochberg adjustment over all properties, no prespecified primary property), **decision models** (Group A diagnostics against usefulness, exploratory, n = 17), **text columns** (refits without one text column's decision block, all seven predictors; no permutation importance, shuffling or SHAP) and **rows** (empty vs long text, baseline uncertainty, decision confidence, model agreement vs disagreement). Finally a fixed rule predicts each dataset's gain in advance (leave one dataset out). Common protocol: seven fixed predictors with library defaults and no tuning (linear/logistic, CatBoost, LightGBM, TabPFN-3.5, TabICL v2, LimiX-2, Kumo); the baseline uses the non-text columns only; 18 datasets with the same five outer folds. Fixed headline metrics: binary ROC-AUC, multiclass macro one-vs-rest ROC-AUC, regression R². Datasets are the unit: per-dataset paired gain, equal-dataset means with dataset-bootstrap intervals, win/tie/loss counts, raw gain and share of headroom. 

Tables come from `output_JEVPFN/exp08_when_text_helps/results/group_b/b3_when_and_for_whom.zip`.

```python
import sys
from pathlib import Path
REPO_ROOT = next(p for p in (Path.cwd(), *Path.cwd().parents) if (p / "pyproject.toml").is_file() and (p / "src").is_dir())
if str(REPO_ROOT) not in sys.path:
    sys.path.insert(0, str(REPO_ROOT))
import matplotlib.pyplot as plt
from src.utils.notebook_display import display_local, display_json, display_table
from src.utils.notebook_report import begin_report, report_section, record, finish_report
from src.utils.provenance import provenance
from src.evaluation import benchmark_export, study_export
from src.visualize import figures, group_b as gbfig, style
style.apply()
NOTEBOOK = "exp08_when_text_helps"
begin_report()
save = figures.FigureSaver(NOTEBOOK)
report_section("1. Scope, provenance and the export read by this notebook")
display_json("Notebook execution identity", provenance())
frames, receipt = benchmark_export.read("B3")
print(study_export.describe(receipt))
record({k: receipt.get(k) for k in ("status", "path", "created_utc", "origin", "sha256", "bytes", "schema", "receipt_sha256")}, "Export identity")
WAITING = "Waiting for export: no measurements are shown."
```

## Text columns: refits without one column

```python
report_section("5. Text columns: refits without one column")
if not frames or frames["column_importance"].empty:
    print(WAITING if not frames else "Text-column refits are pending in this export.")
else:
    importance = frames["column_importance"]
    table = importance.groupby(["dataset_id", "text_column", "model_id"], as_index=False).importance.median()
    display_table("Median over decision models of per-column arm minus its refit without the column", table, preview_rows=14)
    record(table, "Text-column importance (median over decision models)")
```

## Predicting the gain in advance

A ridge rule on two prespecified label-free properties (log median text length, decision redundancy), fitted without the held-out dataset, predicts the gain of the median decision model's both-view features. Reported: rank correlation, use/don't-use accuracy, regret against always and never using text, and a shuffle null.

```python
report_section("7. Predicting the gain in advance")
if not frames or frames["lodo"].empty:
    print(WAITING if not frames else "Leave-one-dataset-out results are pending in this export.")
else:
    display_table("Leave-one-dataset-out evaluation per predictor", frames["lodo"])
    record(frames["lodo"], "LODO evaluation")
    record(frames["lodo_predictions"], "LODO predictions")
    gbfig.render(save, gbfig.lodo_scatter(frames["lodo_predictions"], name="b3_lodo", what="Predicting the gain in advance"))
```
