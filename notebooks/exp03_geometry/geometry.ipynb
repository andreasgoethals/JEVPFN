```python
import sys
from pathlib import Path
REPO_ROOT = next(p for p in (Path.cwd(), *Path.cwd().parents) if (p / "pyproject.toml").is_file() and (p / "src").is_dir())
if str(REPO_ROOT) not in sys.path:
    sys.path.insert(0, str(REPO_ROOT))
import matplotlib.pyplot as plt
from src.utils.notebook_display import display_local, display_json, display_table
from src.utils.notebook_report import begin_report, report_section, record, finish_report
from src.utils.provenance import provenance
from src.visualize import figures, style
style.apply()
NOTEBOOK = "exp03_geometry/geometry"
begin_report()
save = figures.FigureSaver(NOTEBOOK)
report_section("1. Scope, provenance and saved evidence")
display_json("Notebook execution identity", provenance())
```

```python
from src.evaluation.study_reports import encoder_references, geometry_tables
report_section("2. Do neighbours in embedding space receive similar decisions?")
relations, coverage = geometry_tables()
display_table("Available datasets and missing providers", coverage)
display_table("Reference distances between frozen encoders on the same field samples", encoder_references())
if relations.empty:
    record("No measured RQ1C comparisons are present. Each field contributes up to 500 distinct texts.")
else:
    display_table("Observed similarity, permuted reference and neighbourhood overlap", relations)
    by_dataset = relations.groupby(["encoder","provider","dataset"])[["spearman","null_mean","neighbour_overlap"]].mean()
    display_table("Equal-weight dataset means across observed fields", by_dataset.groupby(["encoder","provider"]).mean())
    display_table("Datasets with defined distance correlations", by_dataset.spearman.groupby(["encoder","provider"]).count().to_frame("datasets"))
```
