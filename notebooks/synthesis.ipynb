# Synthesis: statistical framework across experiments

The cross-experiment statistical framework and confirmatory family of `config/synthesis/analysis.yaml`. Datasets are the unit (all 18: 9 classification, 9 regression); folds only form each dataset's five-fold mean. Confirmatory scores shared by both task types: the headroom-normalised gain `(augmented - baseline) / (1 - baseline)` over the same predictor's no-text baseline (B and C families; ROC-AUC, macro one-vs-rest ROC-AUC or R² by task) and the proper-score skill (A2). Pairwise claims use two-sided Wilcoxon signed-rank tests with Holm over the whole declared family; mean ranks and the Nemenyi critical difference are descriptive.
