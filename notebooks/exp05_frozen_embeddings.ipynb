# B2a: Representations without labels

Frozen text embeddings (E5-small-v2, E5-large-v2, Qwen3-Embedding-0.6B; PCA fitted on outer-training rows only: 32 per text column or 64 joint) and the combination of each decision model's both-view features with E5-small-v2 per-column PCA32, compared with the decision features alone. Common protocol: seven fixed predictors with library defaults and no tuning (linear/logistic, CatBoost, LightGBM, TabPFN-3.5, TabICL v2, LimiX-2, Kumo); the baseline uses the non-text columns only; 18 datasets with the same five outer folds. Fixed headline metrics: binary ROC-AUC, multiclass macro one-vs-rest ROC-AUC, regression R². Datasets are the unit: per-dataset paired gain, equal-dataset means with dataset-bootstrap intervals, win/tie/loss counts, raw gain and share of headroom. 

Tables come from `output_JEVPFN/exp05_frozen_embeddings/results/group_b/b2_representations.zip` (`scripts/slurm/rq2_group_b_export.slurm B2`); the label-trained half is in `exp06_tar`.

```python
import sys
from pathlib import Path
REPO_ROOT = next(p for p in (Path.cwd(), *Path.cwd().parents) if (p / "pyproject.toml").is_file() and (p / "src").is_dir())
if str(REPO_ROOT) not in sys.path:
    sys.path.insert(0, str(REPO_ROOT))
import matplotlib.pyplot as plt
from src.utils.notebook_display import display_local, display_json, display_table
from src.utils.notebook_report import begin_report, report_section, record, finish_report
from src.utils.provenance import provenance
from src.evaluation import benchmark_export, study_export
from src.visualize import figures, group_b as gbfig, style
style.apply()
NOTEBOOK = "exp05_frozen_embeddings"
begin_report()
save = figures.FigureSaver(NOTEBOOK)
report_section("1. Scope, provenance and the export read by this notebook")
display_json("Notebook execution identity", provenance())
frames, receipt = benchmark_export.read("B2")
print(study_export.describe(receipt))
record({k: receipt.get(k) for k in ("status", "path", "created_utc", "origin", "sha256", "bytes", "schema", "receipt_sha256")}, "Export identity")
WAITING = "Waiting for export: no measurements are shown."
```

## Coverage

```python
report_section("2. Coverage: planned and complete arms")
if not frames:
    print(WAITING)
else:
    display_table("Planned versus complete five-fold arms per phase and predictor", frames["coverage"], preview_rows=14)
    record(frames["coverage"], "Coverage")
```

## Frozen embeddings and decision + embedding

```python
report_section("3. Frozen embeddings and decision + embedding")
if not frames:
    print(WAITING)
else:
    summary = frames["summary"]
    medians = benchmark_export.median_summary(frames)
    rows = sorted(r for r in summary.representation.unique() if r.startswith("frozen/"))
    rows += ["decision/both/median", "decision+frozen/both/median"]
    table = medians[medians.representation.isin(rows)]
    display_table("Equal-dataset gains of frozen, decision and combined arms", table, preview_rows=21)
    record(table, "B2a summary")
    gbfig.render(save, gbfig.gain_forest(table, rows, name="b2a_gain", what="Representations without labels"))
    contrasts = frames["contrasts"]
    contrasts = contrasts[contrasts.family.isin(["frozen", "decision+frozen"])]
    display_table("Median decision model (both) minus each arm, per dataset and predictor", contrasts, preview_rows=12)
    record(contrasts, "B2a contrasts per dataset")
```
